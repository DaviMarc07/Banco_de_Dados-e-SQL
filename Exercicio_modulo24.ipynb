```python
import sqlite3 # Importando a biblioteca necessária
import pandas as pd

df_vendas = pd.read_csv("TB_VENDAS_TAREFA.csv", delimiter=';') # Carrega a base

conn = sqlite3.connect(':memory:') # Conecta ao BD em memória
df_vendas.to_sql('tb_vendas', conn, index=False, if_exists='replace') # Carrega a tabela

def run_query(query): # Função para executar queries
    return pd.read_sql_query(query, conn)

query = "SELECT * FROM tb_vendas"
result_df = run_query(query)
print(result_df)

query = "SELECT PRODUTO FROM tb_vendas LIMIT 10"
result_df = run_query(query)
print(result_df)

query = """
SELECT 
    AVG(CAST(REPLACE(VALOR_UNID, ',', '.') AS REAL)) AS Media_Valor_Unidade, 
    AVG(UNIDADES) AS Media_Unidades_Vendidas 
FROM tb_vendas
"""
result_df = run_query(query)
print(result_df)

query = """
SELECT 
    ID_COMPRA, 
    ID_CLIENTE, 
    CAST(REPLACE(VALOR_UNID, ',', '.') AS REAL) * UNIDADES AS VALOR_TOTAL_GASTO 
FROM tb_vendas
"""
result_df = run_query(query)
print(result_df)

query = """
SELECT 
    AVG(VALOR_TOTAL_GASTO) AS Media_Total_Geral
FROM 
    (SELECT 
        CAST(REPLACE(VALOR_UNID, ',', '.') AS REAL) * UNIDADES AS VALOR_TOTAL_GASTO 
    FROM tb_vendas)
"""
result_df = run_query(query)
# Exibir o resultado
print(result_df)

# 6)
O **SQL** é primariamente utilizado para as etapas iniciais de extração, filtragem e agregação de **grandes volumes de dados** diretamente na fonte (bancos de dados ou *data warehouses*), garantindo eficiência e integridade no *data wrangling* inicial. Já o **Python** é a linguagem completa usada nas etapas posteriores, sendo essencial para **modelagem preditiva** (Machine Learning), **análise estatística avançada**, visualização complexa de dados e automação de fluxos de trabalho (*pipelines* de dados).

#7)
Bancos de dados **Relacionais** (SQL) se baseiam em uma **estrutura tabular rígida** com schemas pré-definidos (linhas e colunas), sendo ideais para dados estruturados e transações que exigem alta integridade (ACID). Em contraste, bancos **Não Relacionais** (NoSQL) utilizam modelos flexíveis (documentos, chave-valor ou grafos), o que os torna mais adequados para **escalabilidade horizontal**, grandes volumes de dados não estruturados ou semi-estruturados e ambientes que requerem rápidas mudanças no modelo de dados.

#8) 
**Sim, o SQL costuma ser mais prático e eficiente para consultas rápidas** e transformações simples que envolvem dados já armazenados em um SGBD. Isso ocorre porque o SQL é uma **linguagem declarativa**, focada em **"o quê"** queremos, com sintaxe concisa para filtragem e agrupamento. Além disso, a consulta é executada diretamente pelo motor otimizado do banco de dados, o que minimiza a transferência de dados e o uso de memória local, tornando-o mais rápido e conciso para *queries* simples em grandes *datasets* do que o uso do Pandas/Python.
```
